```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("adaniports.csv")
```

```python
df["Trades"] = df["Trades"].fillna(df["Trades"].median())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.describe()
```

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
```

```python
df["Date"].dtype
```

```python
plt.figure(figsize=(12,6))

plt.plot(df["Date"], df["Close"])

plt.title("Closing Price Trend")
plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,6))

plt.plot(df["Date"], df["Volume"])

plt.title("Trading Volume Trend")
plt.xlabel("Date")
plt.ylabel("Volume")

plt.xticks(rotation=45)

plt.show()
```

```python
df["MA20"] = df["Close"].rolling(window=20).mean()
```

```python
df[["Date", "Close", "MA20"]].head(25)
```

```python
plt.figure(figsize=(12,6))

plt.plot(df["Date"], df["Close"], label="Closing Price")
plt.plot(df["Date"], df["MA20"], label="20-Day Moving Average")

plt.title("Closing Price vs 20-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("Price")

plt.legend()

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Close"], bins=30, kde=True)

plt.title("Distribution of Closing Price")
plt.xlabel("Closing Price")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df["Close"])

plt.title("Box Plot of Closing Price")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.select_dtypes(include=np.number).corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
top10 = df.nlargest(10, "Close")

plt.figure(figsize=(10,5))

plt.bar(top10["Date"].astype(str), top10["Close"])

plt.title("Top 10 Highest Closing Prices")
plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.xticks(rotation=45)

plt.show()
```

```python
sns.pairplot(df[["Open","High","Low","Close"]])

plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18,10))

# 1. Closing Price Trend
axes[0,0].plot(df["Date"], df["Close"])
axes[0,0].set_title("Closing Price Trend")

# 2. Trading Volume Trend
axes[0,1].plot(df["Date"], df["Volume"])
axes[0,1].set_title("Trading Volume Trend")

# 3. Histogram
axes[0,2].hist(df["Close"], bins=30)
axes[0,2].set_title("Closing Price Distribution")

# 4. Box Plot
sns.boxplot(y=df["Close"], ax=axes[1,0])
axes[1,0].set_title("Box Plot of Closing Price")

# 5. Bar Chart (Top 10 Highest Closing Prices)
top10 = df.nlargest(10, "Close")
axes[1,1].bar(top10["Date"].astype(str), top10["Close"])
axes[1,1].set_title("Top 10 Highest Closing Prices")
axes[1,1].tick_params(axis='x', rotation=45)

# 6. Scatter Plot (Open vs Close)
axes[1,2].scatter(df["Open"], df["Close"])
axes[1,2].set_title("Open vs Close Price")
axes[1,2].set_xlabel("Open")
axes[1,2].set_ylabel("Close")

plt.tight_layout()
plt.show()
```

### Final Observations

* **Dataset Size**: Contains 3322 records and 15 columns of Adani Ports historical stock data.
* **Data Cleaning**: Missing values in the *Trades* column were filled using the median.
* **Data Integrity**: Zero duplicate records were found in the dataset.
* **Price Trends**: Closing prices fluctuated over time, reflecting changing market trends.
* **Market Activity**: Trading volume varied significantly across different trading days.
* **Moving Average**: The 20-day moving average smoothed short-term fluctuations effectively.
* **Data Distribution**: Histogram and box plots highlighted the closing price distribution and outliers.
* **Price Relationship**: Scatter plot showed a positive relationship between Open and Close prices.
* **Core Correlation**: Heatmap revealed strong positive correlation among Open, High, Low, Close, and VWAP.
* **Conclusion**: Visualizations provided meaningful insights into historical performance and market behavior.


## 💡 Key Insight

The **20-day moving average** consistently smoothed out daily price volatility, revealing an underlying trend direction that raw closing prices alone couldn't show — this confirms why moving averages are a core tool in technical trading analysis.

Additionally, the **strong positive correlation** among `Open`, `High`, `Low`, `Close`, and `VWAP` shows that Adani Ports' stock moved in a fairly **stable, predictable price band** each day.

> 🔑 **Takeaway:** This pricing behavior is typical of a **large-cap, actively traded stock**, not a highly speculative one.
